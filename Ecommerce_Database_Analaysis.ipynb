```python
#Calculating the order_month,order_Year, and Order_Day_Of_week
data['Order_Month']=data['Order Date'].dt.month
data['Order_Year']=data['Order Date'].dt.year
data['Order_Day_Of_Week']=data['Order Date'].dt.dayofweek
```

```python
data.head()
```

```python
sales_by_month=data.groupby('Order_Month')['Sales'].sum().reset_index()
sales_by_month
```

```python
fig=px.line(sales_by_month,x='Order_Month',y='Sales',title='Monthly Sales Analysis')
fig.show()
#November month has the highest sales and February month has the lowest sales
```

```python
#Sales by Category
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
fig=px.pie(sales_by_category,values='Sales',names='Category',title='Sales by Category')
fig.show()
# Technology category has the highest sales
```

```python
sales_by_subCategory=data.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_subCategory
```

```python
#plot a graph for this dtaaframe
fig=px.bar(sales_by_subCategory,x='Sub-Category',y='Sales',title='Sales by Sub-Category')
fig.show()
```

```python
#profit_by_month
profit_by_month=data.groupby('Order_Month')['Profit'].sum().reset_index()
profit_by_month
```

```python
fig=px.line(profit_by_month,x='Order_Month',y='Profit',title='Profit by Month',color_discrete_sequence=px.colors.qualitative.Bold)
fig.show()
```

```python
profit_by_category=data.groupby('Category')['Profit'].sum().reset_index()
profit_by_category
```

```python
fig=px.pie(profit_by_category,values='Profit',names='Category',title='Profit by Category',color_discrete_sequence=px.colors.qualitative.Pastel,hole=0.5)
fig.show()
```

```python
#profit_by_sub_category
profit_by_subcategory=data.groupby('Sub-Category')['Profit'].sum().reset_index()
profit_by_subcategory
```

```python
fig=px.bar(profit_by_subcategory,x='Sub-Category',y='Profit',title='Profit by Sub-Category',color_discrete_sequence=px.colors.qualitative.Pastel)
fig.show()
#
```

```python
#Sales and profit by customer segment
sales_by_customer_segment=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_by_customer_segment
```

```python
fig=go.Figure()
fig.add_trace(go.Bar(x=sales_by_customer_segment['Segment'],y=sales_by_customer_segment['Sales'],name='Sales'))
fig.add_trace(go.Bar(x=sales_by_customer_segment['Segment'],y=sales_by_customer_segment['Profit'],name='Profit'))
fig.update_layout(title='Sales and Profit by Customer Segment',barmode='group')
fig.show()
```

```python
sales_by_customer_segment=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_by_customer_segment['Sales_to_profit_Ratio']=sales_by_customer_segment['Sales']/sales_by_customer_segment['Profit']
sales_by_customer_segment
```

```python
fig=px.bar(sales_by_customer_segment,x='Segment',y='Sales_to_profit_Ratio',color='Segment',title='Sales to Profit Ratio by Customer Segment')
fig.show()
```
